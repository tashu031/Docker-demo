```python
# total sales amount generated by each product category
df2.createOrReplaceTempView("products")
df3.createOrReplaceTempView("orders")
df4.createOrReplaceTempView("order_items")
df5.createOrReplaceTempView("returns")
```

```python
total_sales=spark.sql('''
select p.category , sum(oi.selling_price) as Total_sales from products p inner join order_items oi on p.product_id = oi.product_id
group by p.category ''')
total_sales.show()
```

```python
top_customers=spark.sql('''
select o.customer_id , sum(oi.selling_price) as total_purchase from orders o inner join order_items oi on o.order_id = oi.order_id
group by o.customer_id order by total_purchase desc
''')
top_customers.show(10)
```
